```python
import json
import pandas as pd

# 1. Configuración de rutas y nombres de columnas
ruta_json_entrada = 'oficios.json'
ruta_excel = 'consulta.xlsx'
ruta_json_salida = 'datos_filtrados.json'

col_oficio = 'url_archivo'
col_folio = 'folio_sugo'

# 2. Leer el archivo Excel
df_excel = pd.read_excel(ruta_excel, dtype=str)
df_excel.columns = df_excel.columns.str.strip()

# 3. Crear diccionario de búsqueda: { "oficio_autoridad": "FOLIO SUGO" }
# PRO-TIP: Al poner ambas columnas en dropna, descartamos desde el inicio 
# las filas del Excel donde el oficio o el FOLIO SUGO estén en blanco.
df_limpio = df_excel.dropna(subset=[col_oficio, col_folio])

mapa_folios = dict(zip(
    df_limpio[col_oficio].str.strip(), 
    df_limpio[col_folio].str.strip()
))

# 4. Leer el archivo JSON original
with open(ruta_json_entrada, 'r', encoding='utf-8') as f:
    datos_json = json.load(f)

# 5. Filtrar e insertar FOLIO SUGO solo en los que sí se encontraron
datos_encontrados = [] # Lista nueva solo para los resultados exitosos

for objeto in datos_json:
    oficio_actual = objeto.get(col_oficio)
    
    if oficio_actual is not None:
        oficio_limpio = str(oficio_actual).strip()
        folio_encontrado = mapa_folios.get(oficio_limpio)
        
        # Verificamos que sí exista y que no sea una cadena vacía ("")
        if folio_encontrado:
            objeto[col_folio] = folio_encontrado
            datos_encontrados.append(objeto) # Solo se guarda si entró a este if

# 6. Exportar el nuevo archivo JSON únicamente con los encontrados
with open(ruta_json_salida, 'w', encoding='utf-8') as f:
    json.dump(datos_encontrados, f, ensure_ascii=False, indent=4)

print(f"Proceso completado. Se encontraron {len(datos_encontrados)} de {len(datos_json)} registros.")
print(f"Archivo guardado en: {ruta_json_salida}")
```
