```python
import json
import csv

def cruzar_json_csv(ruta_json_entrada, ruta_csv, ruta_json_salida):
    # 1. Leer el CSV y crear un diccionario de búsqueda {url_archivo: folio_sugo}
    mapa_folios = {}
    with open(ruta_csv, mode='r', encoding='utf-8-sig') as archivo_csv:
        lector = csv.DictReader(archivo_csv)
        for fila in lector:
            # Limpiamos espacios en blanco accidentales
            url = fila.get('url_archivo', '').strip()
            folio = fila.get('folio_sugo', '').strip()
            if url:
                mapa_folios[url] = folio

    # 2. Leer el archivo JSON original
    with open(ruta_json_entrada, mode='r', encoding='utf-8') as archivo_json:
        datos_json = json.load(archivo_json)

    # Función auxiliar para actualizar un objeto individual
    def actualizar_registro(registro):
        if isinstance(registro, dict) and 'url_archivo' in registro:
            url_buscada = str(registro['url_archivo']).strip()
            # Asigna el folio_sugo si encuentra coincidencia, o None (null en JSON) si no existe
            registro['folio_sugo'] = mapa_folios.get(url_buscada, None)

    # 3. Realizar el cruce (soporta tanto una lista de objetos como un único objeto JSON)
    if isinstance(datos_json, list):
        for item in datos_json:
            actualizar_registro(item)
    elif isinstance(datos_json, dict):
        actualizar_registro(datos_json)

    # 4. Guardar el resultado en un nuevo archivo JSON
    with open(ruta_json_salida, mode='w', encoding='utf-8') as archivo_salida:
        json.dump(datos_json, archivo_salida, indent=4, ensure_ascii=False)

    print(f"Cruce completado. Archivo generado en: {ruta_json_salida}")

# --- Ejemplo de uso ---
if __name__ == '__main__':
    # Reemplaza con los nombres reales de tus archivos
    ARCHIVO_JSON_ENTRADA = '../oficios.json'
    ARCHIVO_CSV = '../resultados_temp.csv'
    ARCHIVO_JSON_SALIDA = 'datos_actualizados.json'

    cruzar_json_csv(ARCHIVO_JSON_ENTRADA, ARCHIVO_CSV, ARCHIVO_JSON_SALIDA)
```
